# Community Notes Bot — Supabase Data Explorer

Connects to Supabase and shows table sizes + sample rows for each table.

```python
import os
from dotenv import load_dotenv
from supabase import create_client
import pandas as pd

# Load env from the project root
load_dotenv("../.env.local")

url = os.environ["SUPABASE_URL"]
key = os.environ["SUPABASE_SERVICE_KEY"]
supabase = create_client(url, key)
print(f"Connected to {url}")
```

```python
TABLES = [
    "notes",
    "canonical_note_information",
    "scraped_notewriter_snapshots",
    "pipeline_runs",
    "pipeline_scores",
    "notewriters",
    "bot_configs",
    "public_data_snapshots",
]
```

```python
# Table sizes
sizes = []
for table in TABLES:
    try:
        resp = supabase.table(table).select("*", count="exact").limit(0).execute()
        sizes.append({"table": table, "row_count": resp.count})
    except Exception as e:
        sizes.append({"table": table, "row_count": f"ERROR: {e}"})

pd.DataFrame(sizes)
```

```python
# Sample rows from each table
SAMPLE_SIZE = 3
pd.set_option("display.max_columns", None)

for table in TABLES:
    print(f"\n{'='*60}")
    print(f"  {table}")
    print(f"{'='*60}")
    try:
        resp = supabase.table(table).select("*").limit(SAMPLE_SIZE).execute()
        df = pd.DataFrame(resp.data)
        if df.empty:
            print("  (empty table)")
        else:
            display(df)
    except Exception as e:
        print(f"  ERROR: {e}")
```

```python
# Load CN public data
notes_cn = pd.read_csv("notes-00001.tsv", sep="\t", dtype=str)
status_history = pd.read_csv("noteStatusHistory-00000.tsv", sep="\t", dtype=str)

print(f"CN notes: {len(notes_cn)} rows")
print(f"Status history: {len(status_history)} rows")
```

```python
# Fetch all pipeline_runs from Supabase (paginate to get past 1000 limit)
all_runs = []
page_size = 1000
offset = 0
while True:
    resp = supabase.table("pipeline_runs").select(
        "note_id, tweet_id, tweet_text, has_photo, has_video"
    ).range(offset, offset + page_size - 1).execute()
    all_runs.extend(resp.data)
    if len(resp.data) < page_size:
        break
    offset += page_size

pipeline_runs = pd.DataFrame(all_runs)
print(f"Pipeline runs fetched: {len(pipeline_runs)}")
pipeline_runs.head(2)
```

```python
# Filter status history to only CURRENTLY_RATED_HELPFUL
helpful_notes = status_history[status_history["lockedStatus"] == "CURRENTLY_RATED_HELPFUL"]
print(f"Helpful notes in status history: {len(helpful_notes)}")

# Join: CN notes + status history (filter to helpful) + pipeline_runs
# CN notes <-> status history on noteId
# CN notes <-> pipeline_runs on noteId = note_id
helpful_cn = notes_cn.merge(helpful_notes[["noteId"]], on="noteId", how="inner")
print(f"CN notes that are helpful: {len(helpful_cn)}")

# Join with pipeline_runs
merged = helpful_cn.merge(
    pipeline_runs,
    left_on="tweetId",
    right_on="tweet_id",
    how="inner",
)
# Sort merged by createdAtMillis descending
merged = merged.sort_values(by="createdAtMillis", ascending=False)
print(f"Matched with pipeline_runs: {len(merged)}")

# Select only the important columns
result = merged[["tweet_text", "summary", "has_photo", "has_video"]].rename(
    columns={"summary": "note_text"}
)
result
```

```python
result.to_csv("helpful_cn_notes.csv", index=False)
```
